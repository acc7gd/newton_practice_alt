```python
# Function to calculate f(x) for function
def func(x):
    """Calculate X^2"""
    return (x + 2) ** 2


# Calculate first derivative"""
def first_deriv(x, diff=0.0001):
    """Calculate first derivative"""
    approx1 = (func(x + diff) - func(x)) / diff
    return approx1


# Calculate second derivative
def second_deriv(x, diff=0.0001):
    """Calculate second derivative"""
    approx2 = (first_deriv(x + diff) - first_deriv(x)) / diff
    return approx2


print(func(5))
print(first_deriv(5))
print(second_deriv(5))
```

```python
# Calculate Newton function
def newton(x, stop_threshold, diff=0.0001):
    """Calculate Newton function"""
    subtract = stop_threshold + 1
    while abs(subtract) > stop_threshold:
        x_new = x - first_deriv(x, diff) / second_deriv(x, diff)
        subtract = x_new - x
        x = x_new
    return x


print(newton(5, 0.0001))
```
